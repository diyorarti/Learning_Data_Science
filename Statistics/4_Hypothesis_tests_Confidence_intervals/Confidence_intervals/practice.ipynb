```python
import numpy as np

N = 1000
 
population = np.random.normal(
    loc=100,   # μ
    scale=15,  # σ
    size=N
) 
```

```python
sample_means = []

n = 25

for i in range(1000):
    sample = np.random.choice(population, size=n, replace=True)
    sample_mean = sample.mean()
    sample_means.append(sample_mean)

np.std(sample_means, ddof=1)
```

```python
pop_std = np.std(population, ddof=1)
pop_std / np.sqrt(n)
```
