```python
import numpy as np

from scipy.stats import norm
```

# ONE-SAMPLE Z CONFIDENCE INTERVAL

**Business problem**
- A delivery company wants to estimate `the average delivery time` of its orders. 
---
Assumptions:
- Sample mean:               `32 minutes`
- Population SD (known):     `12 minutes`
- Sample size:               `144`
- Confidence level:          `95%`
- Hypothesis:                `two-sided test`
--- 
One-sample Confidence interval:
- Formula: `CI =  x̄ ± z* · σ / √n`

```python
# assumptions
delivery_sample_mean = 32
delivery_std = 12
delivery_sample_size = 144
delivery_alpha = 1 - 0.95  

# calculating two-sided test z critical value
delivery_z_critical_value = norm.ppf(1 - delivery_alpha / 2)
print(f"Z critical value: {delivery_z_critical_value}")

# calculating Standard Error: SE
delivery_se = delivery_std / np.sqrt(delivery_sample_size)
print(f"Standard Error: {delivery_se}")

# calculating marging of error: E
delivery_E = delivery_z_critical_value * delivery_se
print(F"Margin of Error: {delivery_E}")

# calculating 95% CI
delivery_CI_lower_bound = delivery_sample_mean - delivery_E
delivery_CI_upper_bound = delivery_sample_mean + delivery_E
print(f"Delivery 95% Confidence interval lower bound: {delivery_CI_lower_bound} and upper bound: {delivery_CI_upper_bound}")
```

**Interpretation:**
Based on our sample, we estimate that the true population mean delivery time is between `30.04 and 33.96 minutes`, using a `95% confidence interval`. The method used to construct this interval captures the true population mean in 95% of repeated random samples, assuming its conditions hold.
