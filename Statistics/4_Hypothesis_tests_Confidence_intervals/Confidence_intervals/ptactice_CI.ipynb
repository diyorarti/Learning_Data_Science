# ONE-SAMPLE Z CONFIDENCE INTERVAL

**Business problem**
- A delivery company wants to estimate `the average delivery time` of its orders. 
---
Assumptions:
- Sample mean:               `32 minutes`
- Population SD (known):     `12 minutes`
- Sample size:               `144`
- Confidence level:          `95%`
- Hypothesis:                `two-sided test`
--- 
One-sample Confidence interval(known population standard deviation):
- Formula: `CI =  x̄ ± z* · σ / √n`

```python
# assumptions
delivery_sample_mean = 32
delivery_std = 12
delivery_sample_size = 144
delivery_alpha = 1 - 0.95  

# calculating two-sided test z critical value
delivery_z_critical_value = norm.ppf(1 - delivery_alpha / 2)
print(f"Z critical value: {delivery_z_critical_value}")

# calculating Standard Error: SE
delivery_se = delivery_std / np.sqrt(delivery_sample_size)
print(f"Standard Error: {delivery_se}")

# calculating marging of error: E
delivery_E = delivery_z_critical_value * delivery_se
print(F"Margin of Error: {delivery_E}")

# calculating 95% CI
delivery_CI_lower_bound = delivery_sample_mean - delivery_E
delivery_CI_upper_bound = delivery_sample_mean + delivery_E
print(f"Delivery 95% Confidence interval lower bound: {delivery_CI_lower_bound} and upper bound: {delivery_CI_upper_bound}")
CI_95_width = delivery_CI_upper_bound - delivery_CI_lower_bound
```

**Interpretation:**
Based on our sample, we estimate that the true population mean delivery time is between `30.04 and 33.96 minutes`, using a `95% confidence interval`. The method used to construct this interval captures the true population mean in 95% of repeated random samples, assuming its conditions hold.

### Comparing 95% CI and 99% CI

```python
# calculating alpha
delivery_99_CI_alpha = 1 - 0.99 
# calculating two-sided test z critical value
delivery_99_Z_value = norm.ppf(1 - delivery_99_CI_alpha / 2)
# calculating Margin of error 
delivery_99_E = delivery_99_Z_value * delivery_se
# 99% CI
delivery_99_CI_lower_bound = delivery_sample_mean - delivery_99_E
delivery_99_CI_upper_bound = delivery_sample_mean + delivery_99_E

print(f"99% CI Lower bound: {delivery_99_CI_lower_bound}")
print(f"99% CI Upper bound: {delivery_99_CI_upper_bound}")
```

```python
CI_95_width = delivery_CI_upper_bound - delivery_CI_lower_bound
CI_99_width = delivery_99_CI_upper_bound - delivery_99_CI_lower_bound

print(f"95% CI width: {CI_95_width}")
print(f"99% CI width: {CI_99_width}")
```

Based on a sample of 144 orders, the estimated population mean delivery time is 32 minutes. The 95% confidence interval is [30.04, 33.96] minutes, while the 99% confidence interval is [29.42, 34.58] minutes. The 99% interval is wider because achieving a higher confidence level requires a larger critical value and margin of error, assuming the same data and statistical method.


# ONE-SAMPLE t CONFIDENCE INTERVAL

Business Problem
- A company samples 25 customers to estimate `average monthly spending`. The population standard deviation is unknown.
---
Assumptions:
- Sample mean:               `180 PLN`
- Sample SD:                 `40 PLN`
- Sample size:               `25`
- Confidence level:          `95%`
---
One-sample Confidence Interval(Unknown population standard deviation)
- Formual: `CI = x̄ ± t* · s / √n`

```python
# assumptions
monthly_spending_sample_mean = 180 
monthly_spending_std = 40
monthly_spending_sample_size = 25
```

### 95% two-sided CI 

```python
# alpha 
monthly_spending_alpha = 1 - 0.95
print(f"Alpha: {monthly_spending_alpha}")
# calculatinf two-sided t-critical value
monthly_spending_sample_df = monthly_spending_sample_size -1 
monthly_spendig_t_critical_value = t.ppf(1 - monthly_spending_alpha / 2, df=monthly_spending_sample_df)
print(f"t* critical value: {monthly_spendig_t_critical_value}")
# calculating Standard Error: SE
monthly_spedning_SE = monthly_spending_std / np.sqrt(monthly_spending_sample_size)
print(f"Standard Error: {monthly_spedning_SE}")
# calculating Margin of Error: E
monthly_spending_E = monthly_spendig_t_critical_value * monthly_spedning_SE
print(f"Margin of Error: {monthly_spending_E}")
# calculating Lower bound
monthly_spending_CI_lower_bound = monthly_spending_sample_mean - monthly_spending_E
monthly_spending_CI_upper_bound = monthly_spending_sample_mean + monthly_spending_E

print(f"Customers average monthly spedning lower bound: {monthly_spending_CI_lower_bound}")
print(f"Customers average monthly spedning upper bound: {monthly_spending_CI_upper_bound}")
```

based on our sample, we estimate that thre population customer average spending is between `163.5 and 196.5 PLN` using a 95% confidence interval.

### 95% one-tailed CI Lower bound

```python
# alpha 
monthly_spending_alpha = 1 - 0.95
print(f"Alpha: {monthly_spending_alpha}")
# calculatinf two-sided t-critical value
monthly_spending_sample_df = monthly_spending_sample_size -1 
monthly_spendig_t_critical_value = t.ppf(1 - monthly_spending_alpha, df=monthly_spending_sample_df)
print(f"t* critical value: {monthly_spendig_t_critical_value}")
# calculating Standard Error: SE
monthly_spedning_SE = monthly_spending_std / np.sqrt(monthly_spending_sample_size)
print(f"Standard Error: {monthly_spedning_SE}")
# calculating Margin of Error: E
monthly_spending_E = monthly_spendig_t_critical_value * monthly_spedning_SE
print(f"Margin of Error: {monthly_spending_E}")
# calculating Lower bound
monthly_spending_CI_lower_bound = monthly_spending_sample_mean - monthly_spending_E

print(f"Customers average monthly spedning lower bound: {monthly_spending_CI_lower_bound}")
```

# TWO INDEPENDENT MEANS

**Business Problem**
- An e-commerce company compares customer spending across two independently randomized groups. So they want to estimate average customer spending difference between two groups:
---
**Assumptions**
- Control Group
    1. Sample size: `n = 100`
    2. Sample mean: `μ = 200`
    3. Sample Std:  `σ = 40`
- Treatment Group:
    1. Sample size: `n = 120`
    2. Sample mean: `μ = 218`
    3. Sample Std:  `σ = 55`
- Confidence Level: `CL = 95%`
- Hypothesis: `Two-sided test`
---
**Statistical Test**
- Two-independent means (known standard deviations)
- Formulas: 
    1. `CI = (x̄_B - x̄_A) ± t* · √((s_A² / n_A) + (s_B² / n_B))` I use Welch's CI method because population variances are known, and samples variances are different. 
    2. Degrees of freedom: `df = ((s_A² / n_A) + (s_B² / n_B))² / (((s_A² / n_A)² / (n_A - 1)) +((s_B² / n_B)² / (n_B - 1)))`

```python
# assumptions 
control_group_size = 100
control_group_mean = 200
control_group_std = 40
treatment_group_size = 120
treatment_group_mean = 218
treatment_group_std = 55
```

```python
# observed difference between groups
mean_diff = treatment_group_mean - control_group_mean
print(f"Observed difference between two groups means: {mean_diff}")
# calculating alpga
e_commerce_alpha = 1 - 0.95
print(f"Alpha value: {e_commerce_alpha}")
# degrees of freedom
# Variance components
variance_A = control_group_std**2 / control_group_size
variance_B = treatment_group_std**2 / treatment_group_size
# Welch-Satterthwaite degrees of freedom
numerator = (variance_A + variance_B)**2
denominator = (
    variance_A**2 / (control_group_size - 1)
    +
    variance_B**2 / (treatment_group_size - 1)
)
e_commerce_df = numerator / denominator
print(f"Welch degrees of freedom: {e_commerce_df:.4f}")
# calculating z* critical value
e_commerce_z_critical_value = t.ppf(1 - e_commerce_alpha / 2, df=e_commerce_df)
print(f"critical z* value: {e_commerce_z_critical_value}")
# calculating standard error: SE
e_commerce_SE = np.sqrt(
    (control_group_std ** 2 / control_group_size) +
    (treatment_group_std ** 2 / treatment_group_size)
)
print(f"Standard Error: {e_commerce_SE}")
# Calculating Margin of Error: E
e_commerce_E = e_commerce_z_critical_value * e_commerce_SE
print(f"Margin of Error: {e_commerce_E}")
# Calculating bounds
e_commerce_lower_bound = mean_diff - e_commerce_E
e_commerce_upper_bound = mean_diff + e_commerce_E
print(f"Lower Bound: {e_commerce_lower_bound}")
print(f"Upper Bound: {e_commerce_upper_bound}")
# calculating CI width
e_commerce_CI_width = e_commerce_upper_bound - e_commerce_lower_bound
print(f"95% Confidence Interval width: {e_commerce_CI_width}")
```

based on our samples for Control and Treatment groups, customer average spending difference in these two groups is between `5.35 and 30.65 PLN` using `95% Confidence Interval`. Constructed 95% CI excludes zero.

# PAIRED MEANS CONFIDENCE INTERVAL

**Business Problem** 
- a company measures task completion time for the same 10 employees before and after automation. They want to estimate mean differen.
---
**Assumptions**
| Employee | Before (min) | After (min) |
|---------:|-------------:|------------:|
| 1        | 45           | 40          |
| 2        | 50           | 43          |
| 3        | 42           | 40          |
| 4        | 48           | 42          |
| 5        | 55           | 48          |
| 6        | 47           | 43          |
| 7        | 52           | 46          |
| 8        | 44           | 42          |
| 9        | 49           | 45          |
| 10       | 53           | 47          |
- Confidence Level: `95%`
- Two-sided test
**Confidence Interval**
- Paired means Confidence Interval(unknown population variance): `t-based formula: CI = d̄ ± t* · (s_d / √n)`

```python
before_results = np.array([45, 50, 42, 48, 55, 47, 52, 44, 49, 53])
after_results = np.array([40, 43, 40, 42, 48, 43, 46, 42, 45, 47])

# 1. Paired differences
differences = after_results - before_results

# 2. Sample statistics
mean_diff = np.mean(differences)
std_diff = np.std(differences, ddof=1)
n_sample = len(differences)

# 3. Significance level
alpha = 1 - 0.95

# 4. Standard error
SE = std_diff / np.sqrt(n_sample)

# 5. Degrees of freedom
df = n_sample - 1

# 6. Critical t-value
t_critical = t.ppf(1 - alpha / 2, df=df)

# 7. Margin of error
E = t_critical * SE

# 8. Confidence interval
lower_bound = mean_diff - E
upper_bound = mean_diff + E

print(f"Paired differences: {differences}")
print(f"Mean difference: {mean_diff:.4f}")
print(f"Sample SD: {std_diff:.4f}")
print(f"Standard Error: {SE:.4f}")
print(f"Degrees of freedom: {df}")
print(f"Critical t-value: {t_critical:.4f}")
print(f"Margin of Error: {E:.4f}")
print(f"95% CI: [{lower_bound:.4f}, {upper_bound:.4f}]")
```

Based on measurements from 10 employees, automation is associated with an `estimated reduction in average task completion time of 4.9 minutes`. The 95% confidence interval indicates that the population mean reduction is estimated to be between `3.57 and 6.23 minutes`.
