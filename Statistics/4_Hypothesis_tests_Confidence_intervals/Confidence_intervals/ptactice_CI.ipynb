```python
import numpy as np

from scipy.stats import norm, t
```

# ONE-SAMPLE Z CONFIDENCE INTERVAL

**Business problem**
- A delivery company wants to estimate `the average delivery time` of its orders. 
---
Assumptions:
- Sample mean:               `32 minutes`
- Population SD (known):     `12 minutes`
- Sample size:               `144`
- Confidence level:          `95%`
- Hypothesis:                `two-sided test`
--- 
One-sample Confidence interval(known population standard deviation):
- Formula: `CI =  x̄ ± z* · σ / √n`

```python
# assumptions
delivery_sample_mean = 32
delivery_std = 12
delivery_sample_size = 144
delivery_alpha = 1 - 0.95  

# calculating two-sided test z critical value
delivery_z_critical_value = norm.ppf(1 - delivery_alpha / 2)
print(f"Z critical value: {delivery_z_critical_value}")

# calculating Standard Error: SE
delivery_se = delivery_std / np.sqrt(delivery_sample_size)
print(f"Standard Error: {delivery_se}")

# calculating marging of error: E
delivery_E = delivery_z_critical_value * delivery_se
print(F"Margin of Error: {delivery_E}")

# calculating 95% CI
delivery_CI_lower_bound = delivery_sample_mean - delivery_E
delivery_CI_upper_bound = delivery_sample_mean + delivery_E
print(f"Delivery 95% Confidence interval lower bound: {delivery_CI_lower_bound} and upper bound: {delivery_CI_upper_bound}")
CI_95_width = delivery_CI_upper_bound - delivery_CI_lower_bound
```

**Interpretation:**
Based on our sample, we estimate that the true population mean delivery time is between `30.04 and 33.96 minutes`, using a `95% confidence interval`. The method used to construct this interval captures the true population mean in 95% of repeated random samples, assuming its conditions hold.

### Comparing 95% CI and 99% CI

```python
# calculating alpha
delivery_99_CI_alpha = 1 - 0.99 
# calculating two-sided test z critical value
delivery_99_Z_value = norm.ppf(1 - delivery_99_CI_alpha / 2)
# calculating Margin of error 
delivery_99_E = delivery_99_Z_value * delivery_se
# 99% CI
delivery_99_CI_lower_bound = delivery_sample_mean - delivery_99_E
delivery_99_CI_upper_bound = delivery_sample_mean + delivery_99_E

print(f"99% CI Lower bound: {delivery_99_CI_lower_bound}")
print(f"99% CI Upper bound: {delivery_99_CI_upper_bound}")
```

```python
CI_95_width = delivery_CI_upper_bound - delivery_CI_lower_bound
CI_99_width = delivery_99_CI_upper_bound - delivery_99_CI_lower_bound

print(f"95% CI width: {CI_95_width}")
print(f"99% CI width: {CI_99_width}")
```

Based on a sample of 144 orders, the estimated population mean delivery time is 32 minutes. The 95% confidence interval is [30.04, 33.96] minutes, while the 99% confidence interval is [29.42, 34.58] minutes. The 99% interval is wider because achieving a higher confidence level requires a larger critical value and margin of error, assuming the same data and statistical method.


# ONE-SAMPLE t CONFIDENCE INTERVAL

Business Problem
- A company samples 25 customers to estimate `average monthly spending`. The population standard deviation is unknown.
---
Assumptions:
- Sample mean:               `180 PLN`
- Sample SD:                 `40 PLN`
- Sample size:               `25`
- Confidence level:          `95%`
---
One-sample Confidence Interval(Unknown population standard deviation)
- Formual: `CI = x̄ ± t* · s / √n`

```python
# assumptions
monthly_spending_sample_mean = 180 
monthly_spending_std = 40
monthly_spending_sample_size = 25
```

### 95% two-sided CI 

```python
# alpha 
monthly_spending_alpha = 1 - 0.95
print(f"Alpha: {monthly_spending_alpha}")
# calculatinf two-sided t-critical value
monthly_spending_sample_df = monthly_spending_sample_size -1 
monthly_spendig_t_critical_value = t.ppf(1 - monthly_spending_alpha / 2, df=monthly_spending_sample_df)
print(f"t* critical value: {monthly_spendig_t_critical_value}")
# calculating Standard Error: SE
monthly_spedning_SE = monthly_spending_std / np.sqrt(monthly_spending_sample_size)
print(f"Standard Error: {monthly_spedning_SE}")
# calculating Margin of Error: E
monthly_spending_E = monthly_spendig_t_critical_value * monthly_spedning_SE
print(f"Margin of Error: {monthly_spending_E}")
# calculating Lower bound
monthly_spending_CI_lower_bound = monthly_spending_sample_mean - monthly_spending_E
monthly_spending_CI_upper_bound = monthly_spending_sample_mean + monthly_spending_E

print(f"Customers average monthly spedning lower bound: {monthly_spending_CI_lower_bound}")
print(f"Customers average monthly spedning upper bound: {monthly_spending_CI_upper_bound}")
```

based on our sample, we estimate that thre population customer average spending is between `163.5 and 196.5 PLN` using a 95% confidence interval.

### 95% one-tailed CI Lower bound

```python
# alpha 
monthly_spending_alpha = 1 - 0.95
print(f"Alpha: {monthly_spending_alpha}")
# calculatinf two-sided t-critical value
monthly_spending_sample_df = monthly_spending_sample_size -1 
monthly_spendig_t_critical_value = t.ppf(1 - monthly_spending_alpha, df=monthly_spending_sample_df)
print(f"t* critical value: {monthly_spendig_t_critical_value}")
# calculating Standard Error: SE
monthly_spedning_SE = monthly_spending_std / np.sqrt(monthly_spending_sample_size)
print(f"Standard Error: {monthly_spedning_SE}")
# calculating Margin of Error: E
monthly_spending_E = monthly_spendig_t_critical_value * monthly_spedning_SE
print(f"Margin of Error: {monthly_spending_E}")
# calculating Lower bound
monthly_spending_CI_lower_bound = monthly_spending_sample_mean - monthly_spending_E

print(f"Customers average monthly spedning lower bound: {monthly_spending_CI_lower_bound}")
```
