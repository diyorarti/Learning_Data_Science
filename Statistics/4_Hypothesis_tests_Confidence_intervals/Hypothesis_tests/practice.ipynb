```python
import numpy as np
import pandas as pd

from scipy.stats import norm
```

# Z-test 

## One-sample Z-test

```python
df = pd.read_csv('z_test_data/01_one_sample_z_mean.csv')
df.head()
```

### Two-tailed test
Hypotheses:
- H₀: μ = 250
- Hₐ: μ ≠ 250

```python
mu_0 = 250
```

```python
# known population stadard deviation
sigma = 12

alpha = 0.05
```

```python
# Sample statistics
sample = df['fill_ml']
n = len(sample)

sample_mean = sample.mean()

print("Sample size:", n)
print("Sample mean:", sample_mean)
```

```python
# standard error 
standard_error = sigma / np.sqrt(n)
print("Standard error:", standard_error)
```

```python
# observed Z-statistic
z_observed = (
    sample_mean - mu_0
) /standard_error

print("Observed Z:", z_observed)
```

```python
# Critical values 
critical_value = norm.ppf(1 - alpha / 2)

lower_critical = -critical_value
upper_critical = critical_value

print("Critical values:",
      lower_critical,
      upper_critical)
```

```python
# P-value
p_value = 2 * norm.sf(abs(z_observed))


print("P-value:", p_value)
```

```python
# Dicision using p-value
if p_value < alpha:
    print("Reject H₀")
else:
    print("Fail to reject H₀")
```

```python
# Decision using critical values

if abs(z_observed) > critical_value:
    print("Reject H₀ using critical-value method")
else:
    print("Fail to reject H₀ using critical-value method")
```
