# Z-test 

## One-sample Z-test

```python
df = pd.read_csv('z_test_data/01_one_sample_z_mean.csv')
df.head()
```

```python
df.shape
```

Population

```python
mean_cup_ml_population = df['fill_ml'].mean()
mean_cup_ml_population
```

```python
std_cup_ml_population = df['fill_ml'].std()
std_cup_ml_population
```

Sample

```python
n = 40

cup_fill_sample = df.sample(
    n=n,
    random_state=42
)

cup_fill_sample.head()
```

```python
mean_cup_ml_sample = cup_fill_sample['fill_ml'].mean()
mean_cup_ml_sample
```
