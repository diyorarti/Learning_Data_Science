```python
import numpy as np
import pandas as pd

from scipy.stats import norm
```

# Z-test 

## One-sample Z-test

Hypotheses:
- H₀: μ = 250
- Hₐ: μ ≠ 250
- population standard deviation: 12

```python
df = pd.read_csv('z_test_data/01_one_sample_z_mean.csv')
df.head()
```

```python
# company claims
mu_0 = 250

# known population stadard deviation
sigma = 12

# Significance level
alpha = 0.05
```

### Two-tailed test

```python
# Sample statistics
sample = df['fill_ml']
# smple length
n = len(sample)
# sample mean
sample_mean = sample.mean()
# standard error 
standard_error = sigma / np.sqrt(n)

print("Sample size:", n)
print("Sample mean:", sample_mean)
print("Standard error:", standard_error)
```

```python
# observed Z-statistic
z_observed = (
    sample_mean - mu_0
) /standard_error

print("Observed Z:", z_observed)
```

```python
# Critical values 
critical_value = norm.ppf(1 - alpha / 2)
lower_critical = -critical_value
upper_critical = critical_value
print("Critical values:",
      lower_critical,
      upper_critical)

# Decision using critical values
if abs(z_observed) > critical_value:
    print("Reject H₀ using critical-value method")
else:
    print("Fail to reject H₀ using critical-value method")
```

```python
# P-value
p_value = 2 * norm.sf(abs(z_observed))
print("P-value:", p_value)
print("Alpha:", alpha)

# Dicision using p-value
if p_value < alpha:
    print("Reject H₀")
else:
    print("Fail to reject H₀")
```

### One-tailed test

```python
# Sample statistics
sample = df['fill_ml']
# smple length
n = len(sample)
# sample mean
sample_mean = sample.mean()
# standard error 
standard_error = sigma / np.sqrt(n)

print("Sample size:", n)
print("Sample mean:", sample_mean)
print("Standard error:", standard_error)
```

```python
# observed Z-statistic
z_observed = (
    sample_mean - mu_0
) /standard_error

print("Observed Z:", z_observed)
```

#### Right-tailed test

```python
# calculating critical value
critical_value = norm.ppf(1 - alpha)
print(f"Critical value (Boundary): {critical_value}")

# Decision using critical values
if z_observed > critical_value:
    print("Reject H₀ using critical-value method")
else:
    print("Fail to reject H₀ using critical-value method")
```

```python
# p-value
p_value = 1 - norm.cdf(z_observed)
print(f"P-value: {p_value}")
print(f"alpha: {alpha}")

# Dicision using p-value
if p_value > alpha:
    print("Reject H₀")
else:
    print("Fail to reject H₀")
```

#### Left-tailed test 

```python
# calculating critical value
critical_value = norm.ppf(alpha)
print(f"Critical value (Boundary): {critical_value}")

# Decision using critical values
if z_observed < critical_value:
    print("Reject H₀")
else:
    print("Fail to reject H₀")
```

```python
# p-value
p_value = norm.cdf(z_observed)
print(f"P-value: {p_value}")
print(f"alpha: {alpha}")

# Dicision using p-value
if p_value > alpha:
    print("Reject H₀")
else:
    print("Fail to reject H₀")
```
