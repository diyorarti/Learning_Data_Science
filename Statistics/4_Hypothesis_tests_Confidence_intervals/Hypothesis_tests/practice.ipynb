```python
import numpy as np
import pandas as pd

from scipy.stats import norm
```

# Z-test 

## One-sample Z-test

Hypotheses:
- H₀: μ = 250
- Hₐ: μ ≠ 250
- population standard deviation: 12

```python
df = pd.read_csv('z_test_data/01_one_sample_z_mean.csv')
df.head()
```

```python
# company claims
mu_0 = 250

# known population stadard deviation
sigma = 12

# Significance level
alpha = 0.05
```

### Two-tailed test

```python
# Sample statistics
sample = df['fill_ml']
# smple length
n = len(sample)
# sample mean
sample_mean = sample.mean()
# standard error 
standard_error = sigma / np.sqrt(n)

print("Sample size:", n)
print("Sample mean:", sample_mean)
print("Standard error:", standard_error)
```

```python
# observed Z-statistic
z_observed = (
    sample_mean - mu_0
) /standard_error

print("Observed Z:", z_observed)
```

```python
# Critical values 
critical_value = norm.ppf(1 - alpha / 2)
lower_critical = -critical_value
upper_critical = critical_value
print("Critical values:",
      lower_critical,
      upper_critical)

# Decision using critical values
if z_observed < lower_critical or z_observed > upper_critical:
    print("Reject H₀ using critical-value method")
else:
    print("Fail to reject H₀ using critical-value method")
```

```python
# P-value
p_value = 2 * norm.sf(abs(z_observed))
print("P-value:", p_value)
print("Alpha:", alpha)

# Dicision using p-value
if p_value < alpha:
    print("Reject H₀")
else:
    print("Fail to reject H₀")
```

### One-tailed test

```python
# Sample statistics
sample = df['fill_ml']
# smple length
n = len(sample)
# sample mean
sample_mean = sample.mean()
# standard error 
standard_error = sigma / np.sqrt(n)

print("Sample size:", n)
print("Sample mean:", sample_mean)
print("Standard error:", standard_error)
```

```python
# observed Z-statistic
z_observed = (
    sample_mean - mu_0
) /standard_error

print("Observed Z:", z_observed)
```

#### Right-tailed test

```python
# calculating critical value
critical_value = norm.ppf(1 - alpha)
print(f"Critical value (Boundary): {critical_value}")

# Decision using critical values
if z_observed > critical_value:
    print("Reject H₀ using critical-value method")
else:
    print("Fail to reject H₀ using critical-value method")
```

```python
# p-value
p_value = 1 - norm.cdf(z_observed)
print(f"P-value: {p_value}")
print(f"alpha: {alpha}")

# Dicision using p-value
if p_value > alpha:
    print("Reject H₀")
else:
    print("Fail to reject H₀")
```

#### Left-tailed test 

```python
# calculating critical value
critical_value = norm.ppf(alpha)
print(f"Critical value (Boundary): {critical_value}")

# Decision using critical values
if z_observed < critical_value:
    print("Reject H₀")
else:
    print("Fail to reject H₀")
```

```python
# p-value
p_value = norm.cdf(z_observed)
print(f"P-value: {p_value}")
print(f"alpha: {alpha}")

# Dicision using p-value
if p_value > alpha:
    print("Reject H₀")
else:
    print("Fail to reject H₀")
```

## Two-sample Z-test

Samples:
- X̄_A = 72 
- σ_A = 10
- X̄_B = 76
- σ_B = 12    
------------                     
Hypothesis:
- H₀: μ_A = μ_B
- Hₐ: μ_A ≠ μ_B
- Significance level α = 0.05
------------
Formula:
- Z = ((X̄_A - X̄_B) - (μ_A - μ_B)₀) / √((σ_A² / n_A) + (σ_B² / n_B))

```python
df_two_sample = pd.read_csv("z_test_data/02_two_sample_z_mean.csv")
df_two_sample.head()
```

```python
df_two_sample['group'].value_counts()
```

```python
# sample A 
mu_A = 72
sample_A_std = 10

# sample B 
sample_B_std = 12
mu_B = 76

# Significance level 
alpga = 0.05
```

### Two-tailed Hypothesis 

```python
# sample A 
sample_A = df_two_sample[df_two_sample['group'] == 'A']['satisfaction_score']
sample_A_n = len(sample_A)
sample_A_mean = sample_A.mean()
print("Sample A mean: ", sample_A_mean)

# Sample B
sample_B = df_two_sample[df_two_sample['group'] == 'B']['satisfaction_score']
sample_B_mean = sample_B.mean()
sample_B_n = len(sample_B)
print("Sample B mean: ", sample_B_mean)


```

```python
# Two-sample-Z-statistic calculation: Z = ((X̄_A - X̄_B) - (μ_A - μ_B)₀) / √((σ_A² / n_A) + (σ_B² / n_B))
two_sample_Z_test = (
    ((sample_A_mean - sample_B_mean) - (mu_A - mu_B))
    /
    np.sqrt(
        (sample_A_std ** 2 / sample_A_n) + (sample_B_std ** 2 / sample_B_n)
    )
)

print("Test Statistic Value", two_sample_Z_test)
```
