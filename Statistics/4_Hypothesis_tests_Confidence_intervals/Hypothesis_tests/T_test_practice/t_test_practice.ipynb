```python
import pandas as pd
import numpy as np

from scipy.stats import t
```

# Tests

## One-Sample T-test

A delivery company claims that it's average delivery time `30 minutes`. Population std(σ) is unknown.                                                    
Research questions:
- Is the average delivery time different from 30 minutes
-----
Indentifying the correct Hypothesis test
- I can use `One-Sample T-test` since there is one sample and Standard deviation is unknown
- I choose `Two-tailed One-Sample T-test` because the research questions asked the difference
-----
Hypothesis:
- H₀: μ = 30
- Hₐ: μ ≠ 30
---- 
Significance level α 
- α = 0.05 I accept `5% Type I` error
--- 
Formula:
- t = (X̄ - μ₀) / (s / √n)

```python
delivery_df = pd.read_csv("data/01_one_sample_t_mean.csv")
delivery_df.head()
```

```python
# company claims 
delivery_mu_0 = 30
# significance_level alpha
delivery_alpha = 0.05
```

```python
# sample
delivery_sample = delivery_df['delivery_time_min']
# sample length 
delivery_n = len(delivery_sample)
# degrees of freedom
delivery_df_value = delivery_n - 1
# delivery sample mean 
delivery_mean = np.mean(delivery_sample)
# delivery sample standard deviation
delivery_std = delivery_sample.std()


print(F"Delivery Sample Size: {delivery_n}")
print(F"Delivery Sample Mean: {delivery_mean}")
print(F"Delivery Sample STD: {delivery_std}")
```

```python
# calculating the One-sample T-test
delivery_t_observed = (
    (delivery_mean - delivery_mu_0) 
    /
    (delivery_std / np.sqrt(delivery_n))
)

print(f"Delivery Sample observed One-Sample T-test Value: {delivery_t_observed}")
```

```python
# Finding critical values
delivery_critical_values = t.ppf(1 - delivery_alpha / 2, df=delivery_df_value)
delivery_lower_critical = - delivery_critical_values
delivery_upper_critical = delivery_critical_values
print(F"Delivery Lower Critical Value: {delivery_lower_critical}")
print(F"Delivery Upper Critical Value: {delivery_upper_critical}")

# Making decision by critical values
if delivery_t_observed < delivery_lower_critical or delivery_t_observed > delivery_upper_critical:
    print("Reject H₀")
else:
    print("Fail to Reject H₀")
```

```python
# p-value
delivery_p_value = 2  * t.sf(
    abs(delivery_t_observed), 
    df=delivery_df_value
)
print("Delivery p-value: ", delivery_p_value)
print("Delivery Alpha: ", delivery_alpha)

if delivery_p_value < delivery_alpha:
    print("Reject H₀")
else:
    print("Fail to Reject H₀")
```

At the `5%` significance level, there is sufficient statistical evidence to conclude that the true average delivery time is `different from 30 minutes`.
