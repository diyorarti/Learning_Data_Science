```python
df = pd.read_csv('data.csv')
df.head()
```

```python
X = df.drop(columns=['apartment_price_usd'])
y = df['apartment_price_usd']

n_samples = X.shape[0]

np.random.seed(42)
indices = np.random.permutation(n_samples)

X_shuffled = X.iloc[indices]
y_shuffled = y.iloc[indices]

X_train = X_shuffled[:4000]
y_train = y_shuffled[:4000]

X_test = X_shuffled[4000:]
y_test = y_shuffled[4000:]

print(f"Training Data: {X_train.shape}")
print(f"Training Data Target: {y_train.shape}")
print(f"Test Data: {X_test.shape}")
print(f"Test Data Target: {y_test.shape}")
n_samples
```

```python
def standardization(x):
    mean = x.mean()
    std = x.std()

    x_stand = (x - mean) / std

    return x_stand

X_train_stand = standardization(x=X_train)
X_test_stand = standardization(x=X_test)
```

```python
ones = np.ones((X_train.shape[0], 1))
X_with_intercept = np.hstack((ones, X_train))

beta_hat = np.linalg.pinv(X_with_intercept.T @ X_with_intercept ) @ X_with_intercept.T @ y_train

intercept = beta_hat[0]
coefficients = beta_hat[1:]
```

```python
parameters = pd.DataFrame({
    "Feature": ["Intercept"],
    "Value": [intercept]
})

for feature, coefficient in zip(X.columns, coefficients):
    parameters.loc[len(parameters)] = [feature, coefficient]

parameters
```

```python
train_predictions = X_train @ coefficients + intercept
test_predictions = X_test @ coefficients + intercept
```

# Assessing the model fit

## R2-square

```python
def R2_score(y_true, y_pred):
    if len(y_true) != len(y_pred):
            raise ValueError("The length of y_true and y_pred must be the same.")
    SSR = np.sum((y_true - y_pred)**2)
    SST = np.sum((y_true - np.mean(y_true))** 2)
    return 1 - (SSR / SST)
```

```python
r2_score_train = R2_score(y_train, train_predictions)
r2_score_train
```

```python
r2_score_test = R2_score(y_test, test_predictions)
r2_score_test
```

```python
def adjusted_R2_score(y_true, y_pred, X_train):
    if len(y_true) != len(y_pred):
            raise ValueError("The length of y_true and y_pred must be the same.")

    n, p = X_train.shape
    SSR = np.sum((y_true - y_pred) ** 2)
    SST = np.sum((y_true - np.mean(y_true)) ** 2)
    ordinary_R2 = 1 - (SSR / SST)
    return 1 - (1 - ordinary_R2) * ((n - 1) / (n - p - 1))
```

```python
adj_r2_score_train = adjusted_R2_score(y_train, train_predictions, X_train)
adj_r2_score_train
```

```python
adj_r2_score_test = adjusted_R2_score(y_test, test_predictions, X_test)
adj_r2_score_test
```

## Root mean squared error RMSE

```python
def root_mean_squared_error(y_true, y_pred):
    if len(y_true) != len(y_pred):
                raise ValueError("The length of y_true and y_pred must be the same.")
    
    n = len(y_pred)
    rmse = np.sqrt(np.sum((y_true - y_pred)**2) / n)
    return rmse
```

```python
rmse_train = root_mean_squared_error(y_train, train_predictions)
rmse_train
```

```python
rmse_test = root_mean_squared_error(y_test, test_predictions)
rmse_test
```

## F-statistic

```python
def f_statistic(y_true, y_pred, X_train):
    SSR = np.sum((np.mean(y_true) - y_pred) ** 2)
    SSE = np.sum((y_pred - y_true) ** 2)

    n, p = X_train.shape
    regression_df  = p
    error_of_freedom = n - p - 1

    MSR = SSR / regression_df 
    MSE = SSE / error_of_freedom

    f_statistic = MSR / MSE 
    return f_statistic
```

```python
f_statistic_train = f_statistic(y_train, train_predictions, X_train)
f_statistic_train
```

```python
f_statistic_test = f_statistic(y_test, test_predictions, X_test)
f_statistic_test
```

## AIC

```python
def aic_calculation(y_true, y_pred, X):
    n = len(y_true)
    p = X.shape[1]

    SSE = np.sum((y_true - y_pred) ** 2)

    sigma_squared = SSE / n

    log_likelihood = -(n / 2) * (
        np.log(2 * np.pi)
        + np.log(sigma_squared)
        + 1 
    )

    k = p + 1

    AIC = 2 * k - 2 * log_likelihood
    return AIC
```

## BIC 

```python
def bic_calculation(y_true, y_pred, X):
    n = len(y_true)
    p = X.shape[1]

    SSE = np.sum((y_true - y_pred) ** 2)

    sigma_squared = SSE / n

    log_likelihood = -(n / 2) * (
        np.log(2 * np.pi)
        + np.log(sigma_squared)
        + 1
    )
    k = p + 1

    BIC = k * np.log(n) - 2 * log_likelihood
    return BIC
```
