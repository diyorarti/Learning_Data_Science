## Simple Random Sampling

```python
population_size = len(df)
sample_size = 240

print(f"Population size: {population_size}")
print(f"Sample size: {sample_size}")
```

```python
simple_random_sample = df.sample(
    n=sample_size,
    replace=False,
    random_state=42
)
```

```python
simple_random_sample.head()
```

```python
simple_random_sample.shape
```

```python
simple_random_sample["student_id"].duplicated().sum()
```

Comparing **Population** vs **Sample**

```python
num_columns = ['age', 'study_hours_week', 'attendance_pct', 'exam_score', 'satisfaction_score']
print("Population: ")
df[num_columns].describe()
```

```python
simple_random_sample[num_columns].describe()
```

```python
population_department = (
    df['department'].value_counts(normalize=True).mul(100)
)
sample_depratment = (
    simple_random_sample['department'].value_counts(normalize=True).mul(100)
)
department_comparison = pd.DataFrame({
    "Population Percentage":np.round(population_department,1),
    "Sample Percentage": np.round(sample_depratment,1)
})
department_comparison
```

```python
population_region = (
    df['region'].value_counts(normalize=True).mul(100)
)
sample_region = (
    simple_random_sample['region'].value_counts(normalize=True).mul(100)
)
region_comparison = pd.DataFrame({
    "Population Percentage":np.round(population_region,1),
    "Sample Percentage": np.round(sample_region,1)
})
region_comparison
```

```python
population_gender = (
    df['gender'].value_counts(normalize=True).mul(100)
)
sample_gender = (
    simple_random_sample['gender'].value_counts(normalize=True).mul(100)
)
gender_comparison = pd.DataFrame({
    "Population Percentage":np.round(population_gender,1),
    "Sample Percentage": np.round(sample_gender,1)
})
gender_comparison
```
