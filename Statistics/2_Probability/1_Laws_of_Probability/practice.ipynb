#### Independent vs Dependent Events

Whether the two events are independent or Dependent ?

##### Method-1: 

- When two events are independent P(A ∩ B) = P(A) × P(B)​                                                                                        
- When two events are dependent   P(A ∩ B) ≠ P(A) × P(B)​

```python
event_A = cupon_purchase[cupon_purchase['coupon_used'] == 'Yes'] 
event_B = cupon_purchase[cupon_purchase['purchased'] == 'Yes']
event_A_and_B = cupon_purchase[
    (cupon_purchase['coupon_used'] == 'Yes') &  
    (cupon_purchase['purchased'] == 'Yes')
]

n = len(cupon_purchase)

p_A = len(event_A) / n
p_B = len(event_B) / n
p_A_and_B = len(event_A_and_B) / n

independence_probability = p_A * p_B


print("P(A):", p_A)
print("P(B):", p_B)
print("P(A ∩ B):", p_A_and_B)
print("P(A) × P(B):", independence_probability)

if p_A_and_B == independence_probability:
    print("Events A and B are independent. Because P(A ∩ B) = P(A) * P(B)")
else:
    print("Events A and B are dependent. Because P(A ∩ B) != P(A) * P(B). ")
```

##### Method - 2:

- When two events are independent P(B | A) = P(B)
- When two evets are dependent    P(B | A) ≠ P(B) 

P(B | A) = P (A ∩ B) / P(A)

```python
# events
event_A_B = cupon_purchase[
    (cupon_purchase['coupon_used'] == 'Yes') &  
    (cupon_purchase['purchased'] == 'Yes')
]
event_A = cupon_purchase[cupon_purchase['coupon_used'] == 'Yes']
n = len(cupon_purchase)

# probabilities
p_A =  len(event_A) / n
p_B_given_A = len(event_A_B) / len(event_A)


print(f"Probability of A {p_A}")
print(f"Probability of B given A {p_B_given_A}")


if p_B_given_A == p_B:
    print("Events A and B are independent. Because P(B|A) = P(B)")
else:
    print("Events A and B are dependent. Because P(B|A) != P(B)")
```

#### Addition rule 

##### General/Dependent multiplication rule

P(A U B)= P(A) + P(B) - P(A ∩ B)

```python
# events
evnt_A = cupon_purchase[cupon_purchase['coupon_used'] == 'Yes']
evnt_B = cupon_purchase[cupon_purchase['purchased'] == 'Yes']
n = len(cupon_purchase)

# events probabilities
prob_A = len(evnt_A) / n
prob_B = len(evnt_B) / n


# intersection events probabilities 
inters_A_B = cupon_purchase[
    (cupon_purchase['coupon_used'] == 'Yes') & 
    (cupon_purchase['purchased'] == 'Yes')
]
inters_events_prob = len(inters_A_B) / n

# union events probabilities
union_events_prob = prob_A + prob_B - inters_events_prob


print(f"Probability of event A happening: {prob_A}")
print(f"Probability of event B happening: {prob_B} ")
print(f"Probability of event A and B happening: {inters_events_prob} ")
print(f"Probability of event A or B happening: {union_events_prob} ")
```

#### Conditional Probability

- Event A - Customers used cupon
- Event B - Customers purchased

P(A | B) = P( A ∩ B) / P(B)                                                                                                                                           
P(A | B)  means what is the probability of happening event A when event B already happens


```python
event_A = df[df['coupon_used'] == 'Yes']
event_B = df[df['purchased'] == 'Yes']
n = len(df)

event_A_and_B = df[
    (df['coupon_used'] == 'Yes') &
    (df['purchased'] == 'Yes')
]

prob_A_B = len(event_A_and_B) / n 
prob_B = len(event_B) / n
prob_A_given_B = prob_A_B  / prob_B

print(f"Probability of event A and B: {prob_A_and_B}")
print(f"Probability of event B {prob_B}")
print(f"Probability of event A when B is already happened: {prob_A_given_B}")
```

#### Bayes' Theorem



Bayes' Theorem reverses the direction of probability for example:
Suppose we have `P(B | A)`,                                                                                                                                 
but we want `P(A | B)`,                                                                                                                                             
so Bayes' Theorem: `P(A | B) = P(B | A) * P(A) / P(B)`                                                                                                                          
Probability of A given B not equal to Probability of B given A `P(A | B) ≠ P(B | A)`                                                
- Event A - Customers used cupon
- Event B - Customers purchased

```python
# Events
event_A = df[df["coupon_used"] == "Yes"]
event_B = df[df["purchased"] == "Yes"]

n = len(df)

# Intersection: coupon used AND purchased
event_A_and_B = df[
    (df["coupon_used"] == "Yes") &
    (df["purchased"] == "Yes")
]

# Probabilities
prob_A = len(event_A) / n
prob_B = len(event_B) / n
prob_A_and_B = len(event_A_and_B) / n

# Conditional probability P(B | A)
prob_B_given_A = prob_A_and_B / prob_A

# Bayes' theorem: P(A | B)
prob_A_given_B = (
    prob_B_given_A * prob_A
    / prob_B
)

print(f"Probability of using a coupon, P(A): {prob_A:.4f}")
print(f"Probability of purchasing, P(B): {prob_B:.4f}")
print(f"Probability of coupon usage and purchase, P(A ∩ B): {prob_A_and_B:.4f}")
print(f"Probability of purchasing given coupon usage, P(B | A): {prob_B_given_A:.4f}")
print(f"Probability of coupon usage given purchase, P(A | B): {prob_A_given_B:.4f}")
```
