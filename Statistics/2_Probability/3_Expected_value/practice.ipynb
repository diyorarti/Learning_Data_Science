# First Combination

```python
visits_30d = df['visit_count_30d']

visits_30d.unique()
```

## Expected Value

```python
visit_frequency = visits_30d.value_counts().sort_values()
visit_frequency
```

```python
visit_probability = visit_frequency / len(df) 
visit_probability
```

```python
expected_visit = np.sum(visit_frequency.index * visit_probability.values)
expected_visit
```

```python
df['visit_count_30d'].mean()
```

```python
visits_results = pd.DataFrame(
    {
        "value":visit_frequency.index,
        "probability":visit_probability.values
    }
)
visits_results
```

```python
# expected value of Y 
expected_value_Y = np.sum(visits_results['transformation'] * visits_results['probability'])
expected_value_Y
```

### 4-Propery: Expected value of s sum

`E[X + Y] = E[X] + E[Y]`
- X = premium members visits count
- Y = not premium members visits count
- Z = total visits

```python
X = df.loc[
    df['premium_member'] == 'Yes',
    'visit_count_30d'
]

Y = df.loc[
    df['premium_member'] == 'No',
    'visit_count_30d'
]
```

```python
X_frequency = X.value_counts().sort_values()
X_probability = X_frequency / len(df)
expected_value_X = np.sum(X_frequency.index * X_probability.values)
expected_value_X
```

```python
Y_frequency = Y.value_counts().sort_values()
Y_probability = Y_frequency / len(df)
expected_value_Y = np.sum(Y_frequency.index * Y_probability.values)
expected_value_Y
```

```python
total_visits = expected_value_X + expected_value_Y
total_visits
```

### 6-Property: Multiplication

- if X and Y are independent `E[XY] = E[X]E[Y]`
- if X and Y are dependent `E[XY] ≠ E[X] * E[Y]`
----
- X = visits count of premium users in last 30 days
- Y = visits count of non-premium users in last 30 days

They are `dependet` varianles

```python
X = np.where(
    df["premium_member"] == "Yes",
    df["visit_count_30d"],
    0
)

Y = np.where(
    df["premium_member"] == "No",
    df["visit_count_30d"],
    0
)
```

```python
visits_prem_vs_nonprem = pd.DataFrame({
    "premium_visits":X,
    "non_premium_visits":Y
})
visits_prem_vs_nonprem.head()
```

```python
visits_prem_vs_nonprem['multiplication'] = visits_prem_vs_nonprem['premium_visits'] * visits_prem_vs_nonprem['non_premium_visits']
visits_prem_vs_nonprem.head()
```

```python
X_freq = visits_prem_vs_nonprem['premium_visits'].value_counts()
X_prob = X_freq / len(df)
X_expected_value = np.sum(X_freq.index * X_prob.values)
X_expected_value
```

```python
Y_freq = visits_prem_vs_nonprem['non_premium_visits'].value_counts()
Y_prob = Y_freq / len(df)
Y_expected_value = np.sum(Y_freq.index * Y_prob.values)
Y_expected_value
```

```python
XY_freq = visits_prem_vs_nonprem['multiplication'].value_counts()
XY_prob = XY_freq / len(df)
XY_expected_value = np.sum(XY_freq.index * XY_prob.values)
XY_expected_value
```

```python
X_expected_value * Y_expected_value
```

## Variance

```python
visits_results
```

```python
visits_results['squared_diviations'] = (visits_results['value'] - expected_visit) ** 2 
visits_results
```

```python
visits_variance = np.sum(visits_results['squared_diviations'] * visits_results['probability'])
visits_variance
```

## Standard Deviation

```python
visits_standard_deviation = np.sqrt(visits_variance)
visits_standard_deviation
```

## Covariance

What is the coveriance `Cov(X, Y) = E[(X - E[X]) * (Y - E[Y])]`
- X = visits 
- Y = 2X

```python
visits_results
```

```python
expected_visit
```

```python
expected_visit_of_2X = np.sum(visits_results['2X'] * visits_results['probability'])
expected_visit_of_2X
```

```python
visits_results['(X-E[X])*(Y-E[Y])'] = (visits_results['value'] - expected_visit) * (visits_results['2X'] - expected_visit_of_2X)
visits_results
```

```python
coveriance_X_Y = np.sum(visits_results['probability'] * visits_results['(X-E[X])*(Y-E[Y])'])
coveriance_X_Y
```
