```python
population_customer_spend = df['customer_spend']
population_mean = population_customer_spend.mean()
population_mean
```

```python
n_values = [10, 50, 100, 500, 1000, 5000, 10000]

def sample_mean_calculator(n, population):
    sample_means = pd.DataFrame(columns=['n', 'sample_mean'])
    for n in n_values:
        sample = population.sample(
            n=n,
            replace = True
        )
        sample_mean = sample.mean()
        sample_means.loc[len(sample_means)] = [n, sample_mean]
    return sample_means
```

```python
sample_means_df = sample_mean_calculator(n_values, population_customer_spend)
sample_means_df
```

here we can oberve that as n increase, the sample mean converges the population mean
